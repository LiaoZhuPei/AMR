```python
import os
import onnx
import onnxruntime
from onnxruntime.quantization import quantize_dynamic, QuantType

# 配置路径
RESULTS_DIR = "./training_results_tensor"
INPUT_MODEL = os.path.join(RESULTS_DIR, "best_model.onnx")
OUTPUT_MODEL = os.path.join(RESULTS_DIR, "best_model_int8.onnx")

def main():
    print(f"ONNX Version: {onnx.__version__}")
    print(f"Runtime Version: {onnxruntime.__version__}")

    if not os.path.exists(INPUT_MODEL):
        print("未找到模型文件")
        return

    print("开始量化 (适配 ONNX Runtime 1.8.x)...")
    
    # 1.8.1 版本的 quantize_dynamic 用法
    # 注意：老版本可能不支持 optimize_model=True 参数，或者 weight_type 参数名字不同
    # 这里使用最基础的调用方式，兼容性最强
    
    quantize_dynamic(
        model_input=INPUT_MODEL,
        model_output=OUTPUT_MODEL,
        # 在老版本中，有时需要用 int8 而不是 uint8，或者直接用默认值
        # 如果报错，尝试去掉 weight_type 参数
        weight_type=QuantType.QUInt8 
    )

    print(f"量化完成！输出路径: {OUTPUT_MODEL}")
    print(f"原大小: {os.path.getsize(INPUT_MODEL)/1024:.2f} KB")
    print(f"量化后: {os.path.getsize(OUTPUT_MODEL)/1024:.2f} KB")

if __name__ == "__main__":
    main()
```
