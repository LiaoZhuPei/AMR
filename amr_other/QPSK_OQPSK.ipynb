```python
import matplotlib.pyplot as plt
import numpy as np

# 读取函数 (记得用大端序 >i2)
def check_amplitude(file_path, title):
    raw = np.fromfile(file_path, dtype='>i2')
    # 取前 200 个点足够了，看细节
    samples = raw[1000:1200:2].astype(np.float32) + 1j * raw[1001:1201:2].astype(np.float32)
    
    # 计算幅度
    amplitude = np.abs(samples)
    
    plt.figure(figsize=(10, 4))
    plt.plot(amplitude, '.-', label='Amplitude')
    plt.title(title)
    plt.grid(True)
    # 强制Y轴从0开始，看能不能触底
    plt.ylim(bottom=0) 
    plt.show()

# 请替换为你的真实文件路径
check_amplitude("E:/Project/AMR-Benchmark/QPSK-03-24-25-10h53m20s46.iq", "Real QPSK Amplitude")
# 如果有 OQPSK 文件也画一下
check_amplitude("E:/Project/AMR-Benchmark/OQPSK-03-24-25-10h54m02s6.iq", "Real OQPSK Amplitude")
```
