```python
import torch
from torchvision import models
import torch.nn as nn
import LSTM as mcl

#classes = ['BPSK', 'QPSK', '8PSK', 'OQPSK', '2FSK', '4FSK',
#           '16QAM', 'MSK', 'Pi4DQPSK']

classes = ['BPSK', 'QPSK', '8PSK', 'OQPSK', '2FSK', '4FSK',
           '16QAM', '2ASK', 'Pi4DQPSK', '16APSK', '32APSK','64QAM','32QAM']

model = mcl.LSTM(num_classes=len(classes))
map_location=torch.device('cpu')
device = "cpu"
#weights_path = "E:/translearn/resnet18_premodel_A_rotateMS_8claC_28.pt"
weights_path = "weights/pytorch_lstm4_weights_iq_normalized_13claA.pt"
# 加载模型权重
model.load_state_dict(torch.load(weights_path,map_location=torch.device('cpu')))
model = model.to(device)  


# 模型推理模式
model.eval()
model.cpu()
# 虚拟输入数据
dummy_input1 = torch.randn(1, 1024, 2).to(device)  
# 模型转化函数
torch.onnx.export(model, (dummy_input1), "AMR_2410_LSTM4_0819.onnx", verbose=True, opset_version=13,do_constant_folding=False)
```
