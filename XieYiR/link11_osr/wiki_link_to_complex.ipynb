```python
import wave
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import stft


filename = "E:/XieYiR/wiki_signal/Link-11 CLEW.wav"


# =====================
# 读取WAV
# =====================
f = wave.open(filename)

fs = f.getframerate()

raw = f.readframes(f.getnframes())

data = np.frombuffer(
    raw,
    dtype=np.int16
)

data = data.reshape(-1,2)


# =====================
# I/Q
# =====================
I = data[:,0].astype(float)
Q = data[:,1].astype(float)


# 归一化
I /= 32768
Q /= 32768


# 复信号
z = I + 1j*Q
```

```python
zzz=z.astype(np.complex64)
```

```python
zzz.dtype
```

```python
zzz.shape
```

```python
segment_length = 2048

# 只保留能够完整切分的部分
N = len(zzz) // segment_length
zzz_cut = zzz[:N * segment_length]

# 无重叠切分
zzz_split = zzz_cut.reshape(N, segment_length)

print("切分后的 shape:", zzz_split.shape)
# (352, 2048)

# 保存为 .npy
np.save("zzz_split.npy", zzz_split)

print("已保存：zzz_split.npy")
```
